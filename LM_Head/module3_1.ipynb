```python
from master_llm import MasterLLM
from master_tokenizer import MasterTokenizer
import torch
import torch.nn as nn
```

```python
u_tokenizer = MasterTokenizer("tokenizer.json")

prompt = "the capital of united"

tokens = u_tokenizer.encode(prompt)
```

```python
u_model
```

```python
torch.manual_seed(1)

u_model = MasterLLM(vocab_size=len(u_tokenizer.vocab), embedding_dim=4, num_heads=2, context_length=32, num_layers=3)

out = u_model(tokens)
```

input: [A, B, C]

target: [B, C, D]

```python
import torch

probs=torch.softmax(out[0], dim=-1)
max_probs, max_indice = torch.max(probs, dim=-1)

max_probs, max_indice,probs
```
