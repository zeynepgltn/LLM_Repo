###Model olarak şimdiye kadar yaptığımızı kullanalım

```python
from master_llm import MasterLLM
from master_tokenizer import MasterTokenizer
import torch
```

```python
m_tokenizer = MasterTokenizer("tokenizer.json")
```

```python
prompt = "the capital of united states and the capital of france"
```

```python
tokens = m_tokenizer.encode(prompt)
tokens = torch.tensor(tokens)
```

```python
model = MasterLLM(vocab_size=len(m_tokenizer.vocab), embedding_dim=4, context_length=32)

model(tokens)
```
