```python
from master_llm import MasterLLM
from master_tokenizer import MasterTokenizer
import torch
import torch.nn as nn
```

```python
u_tokenizer = MasterTokenizer("tokenizer.json")

prompt = "the capital of united"

tokens = u_tokenizer.encode(prompt)
```

```python
torch.manual_seed(1)

u_model = MasterLLM(vocab_size=len(u_tokenizer.vocab), embedding_dim=4, num_heads=2, context_length=32)

sentence_meanings_with_attention_context = u_model(tokens)

sentence_meanings_with_attention_context
```

```python
from master_norm import MasterLayerNorm

norm_layer = MasterLayerNorm(embedding_dim=4)
norm_layer(sentence_meanings_with_attention_context)
```
